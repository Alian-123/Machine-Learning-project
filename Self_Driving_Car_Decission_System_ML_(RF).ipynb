```python
# Step 1: Import libraries
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# Step 2: Dataset
data = {
    'Speed': [20, 40, 60, 80, 30, 50, 70, 90],
    'Distance': [100, 80, 60, 40, 90, 70, 50, 30],
    'Action': ['Go', 'Go', 'Slow', 'Stop', 'Go', 'Slow', 'Stop', 'Stop']
}

df = pd.DataFrame(data)

# Step 3: Convert labels to numbers
df['Action'] = df['Action'].map({'Go': 0, 'Slow': 1, 'Stop': 2})

# Step 4: Features & Labels
X = df[['Speed', 'Distance']]
y = df['Action']

# Step 5: Train model
model = RandomForestClassifier(n_estimators=10)
model.fit(X, y)

# Step 6: Take user input
speed = int(input("Enter Speed: "))
distance = int(input("Enter Distance from obstacle: "))

prediction = model.predict([[speed, distance]])

# Step 7: Convert back to text
actions = {0: 'Go 🚗', 1: 'Slow 🐢', 2: 'Stop 🛑'}
print("Car should:", actions[prediction[0]])
```
